# Verify Kaggle output saving before another embedding run
Import this notebook into a new Kaggle notebook. Keep Accelerator set to None. No input dataset is required. Choose Save Version > Save & Run All. After the saved version finishes, open its Output tab and confirm persistence-check/PROOF.json exists, then download and inspect it. This checks output saving only; it does not generate embeddings or validate GPU capacity.

```python
import json
from pathlib import Path
from datetime import datetime, timezone
import uuid

output = Path('/kaggle/working/persistence-check')
output.mkdir(parents=True, exist_ok=True)
proof = {'purpose': 'Verify saved Kaggle output can be reopened', 'run_id': str(uuid.uuid4()), 'created_utc': datetime.now(timezone.utc).isoformat(), 'embeddings_generated': False}
target = output / 'PROOF.json'
target.write_text(json.dumps(proof, indent=2) + '\n', encoding='utf-8')
assert json.loads(target.read_text(encoding='utf-8')) == proof
print('File written:', target)
print(json.dumps(proof, indent=2))
print('Saving is confirmed ONLY after PROOF.json is visible and downloadable in this saved version Output tab.')
```
